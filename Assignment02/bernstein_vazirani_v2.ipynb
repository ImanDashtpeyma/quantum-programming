```python
from qiskit import QuantumCircuit, transpile
from qiskit.providers.basic_provider import BasicSimulator
```

```python
def inner_product(circuit, a):
    n = len(a)
    out = n
    circuit.barrier()
    ra = a[::-1]
    for i,b in enumerate(ra):
        if b=='1': circuit.cx(i,out)
    circuit.barrier()
```

```python
def hadamards(circuit):
    circuit.barrier()
    for q in range(circuit.num_qubits): circuit.h(q)
    circuit.barrier()

def Hadamards(circuit): hadamards(circuit)
```

```python
def bernstein_vazirani(a):
    n = len(a)
    qc = QuantumCircuit(n+1, n)
    out = n
    qc.x(out)
    hadamards(qc)
    inner_product(qc, a)
    hadamards(qc)
    for i in range(n): qc.measure(i,i)
    return qc
```

```python
def run_bv(a, shots=1024):
    qc = bernstein_vazirani(a)
    backend = BasicSimulator()
    tqc = transpile(qc, backend)
    job = backend.run(tqc, shots=shots)
    res = job.result().get_counts()
    return qc, res
```
