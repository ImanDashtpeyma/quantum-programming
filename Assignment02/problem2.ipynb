```python
from qiskit import QuantumCircuit
```

# Step 1: Inverse Product

```python
# Task 1
def inner_product(circuit, a):
    circuit.barrier()
    n = len(a)
    reversed_a = a[::-1]
    for i in range(n):
        if reversed_a[i] == '1':
            circuit.cx(i, n)
    circuit.barrier()
    

# Task 1 Test
def test_circuit_1():
    a = "01101"
    n=len(a)
    circuit = QuantumCircuit(n+1, 0)
    inner_product(circuit, a)
    print(circuit)

test_circuit_1()
```

# Step 2: Hadamard

```python
# Task 2
def hadamards(circuit):
    circuit.barrier()
    num_qubits = circuit.num_qubits
    for qubit in range(num_qubits):
        circuit.h(qubit)
    circuit.barrier()

# Task 2 Test
def test_circuit_2():
    circuit = QuantumCircuit(5, 0)
    hadamards(circuit)
    print(circuit)

test_circuit_2()
```

# Step 3: Bernstein-Vazirani

```python
#Task 3
def bernstein_vazirani(a):
    n = len(a)
    circuit = QuantumCircuit(n + 1, n)
    circuit.x(n)
    hadamards(circuit)
    inner_product(circuit, a)
    hadamards(circuit)
    circuit.measure(range(0,n), range(0,n))
    return circuit

# Task 3 Test
def test_circuit_3():
    a = "01101"
    circuit = bernstein_vazirani(a)
    print(circuit)

test_circuit_3()
```

# Step 4: Simulation Run

```python
from qiskit import transpile
from qiskit_aer import AerSimulator
from qiskit.providers.basic_provider import BasicSimulator

a = "01101"

simulator = AerSimulator()
circuit = bernstein_vazirani(a)
circuit = transpile(circuit, simulator)
result = simulator.run(circuit, shots=10000).result()
counts = result.get_counts(circuit)
print(counts)
```
